#  Exponentially Weighted Moving Average

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
# Load the dataset (added quotes around the file path)
df = pd.read_csv("/home/ipartzix/ML-Dataset/DailyDelhiClimateTest.csv")

# Select the columns (added quotes around column names)
df = df[["date", "meantemp"]]
print(df)

# Convert 'date' to datetime objects so the x-axis formats cleanly
df["date"] = pd.to_datetime(df["date"])
```

```python
# Create and display the scatter plot
plt.scatter(df["date"], df["meantemp"])
plt.xlabel("Date")
plt.ylabel("Mean Temperature")
plt.title("Daily Delhi Climate - Mean Temperature")
plt.show()
```

```python
x1 = df["meantemp"].ewm(alpha=0.9).mean()
print(x1)

df["ewma"] = x1
print(df)

# Ensure x1 is defined (e.g., Exponentially Weighted Moving Average)
# x1 = df['meantemp'].ewm(alpha=0.9).mean()

plt.scatter(df["date"], df["meantemp"], label="Actual Data")
plt.plot(df["date"], x1, color="black", label="alpha = 0.9")  # Added missing comma
plt.xlabel("Date")
plt.ylabel("Mean Temperature")
plt.title("Daily Delhi Climate - Mean Temperature")
plt.legend()  # Displays the labels on the graph
plt.show()
```
