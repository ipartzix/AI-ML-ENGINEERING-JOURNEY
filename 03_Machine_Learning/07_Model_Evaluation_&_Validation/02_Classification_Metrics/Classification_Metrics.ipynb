# Classification Metrics

Classification metrics are used to evaluate classification models.

Examples:

* Logistic Regression
* Decision Trees
* Random Forest
* SVM
* Neural Networks

---

# Confusion Matrix

A confusion matrix is the foundation of most classification metrics.

| Actual / Predicted | Positive | Negative |
| ------------------ | -------- | -------- |
| Positive           | TP       | FN       |
| Negative           | FP       | TN       |

Where:

* TP → True Positive
* TN → True Negative
* FP → False Positive
* FN → False Negative

---

# 1. Accuracy

Measures overall correctness of the model.

## Formula

[
Accuracy = \frac{TP + TN}{TP + TN + FP + FN}
]

## Use Case

* Balanced datasets

## Problem

Not reliable for imbalanced datasets.

Example:

* 95% normal cases
* 5% fraud cases

A model predicting everything as normal gets 95% accuracy but is useless.

---

# 2. Precision

Measures how many predicted positives are actually positive.

## Formula

[
Precision = \frac{TP}{TP + FP}
]

## Focus

False Positives

## Important When

False positives are costly.

Examples:

* Spam detection
* Email filtering

---

# 3. Recall (Sensitivity)

Measures how many actual positives were correctly identified.

## Formula

[
Recall = \frac{TP}{TP + FN}
]

## Focus

False Negatives

## Important When

Missing positive cases is dangerous.

Examples:

* Cancer detection
* Fraud detection

---

# 4. F1 Score

Harmonic mean of Precision and Recall.

## Formula

[
F1 = 2 \times \frac{Precision \times Recall}{Precision + Recall}
]

## Use Case

* Imbalanced datasets
* Need balance between Precision and Recall

---

# 5. Specificity

Measures how well the model identifies negative cases.

## Formula

[
Specificity = \frac{TN}{TN + FP}
]

## Important In

Medical testing

---

# 6. ROC Curve

ROC = Receiver Operating Characteristic Curve

Shows:

* TPR vs FPR

Where:

* TPR = Recall
* FPR = False Positive Rate

---

# 7. ROC-AUC Score

AUC = Area Under Curve

## Range

* 1.0 → Perfect model
* 0.5 → Random guessing
* <0.5 → Bad model

## Use Case

Measures overall classification performance.

---

# 8. Log Loss (Cross Entropy Loss)

Measures prediction uncertainty.

## Formula

[
LogLoss = -\frac{1}{N}\sum[y\log(\hat{y}) + (1-y)\log(1-\hat{y})]
]

## Important For

* Logistic Regression
* Neural Networks

Lower value is better.

---

# Multi-Class Classification Metrics

Used when more than 2 classes exist.

Examples:

* Cat/Dog/Bird classification
* Digit recognition

Techniques:

* Macro Average
* Micro Average
* Weighted Average

---

# Accuracy vs Precision vs Recall

| Metric    | Focus                      | Best Use          |
| --------- | -------------------------- | ----------------- |
| Accuracy  | Overall correctness        | Balanced data     |
| Precision | FP reduction               | Spam detection    |
| Recall    | FN reduction               | Medical diagnosis |
| F1 Score  | Precision + Recall balance | Imbalanced data   |

---

# Important Interview Questions

## 1. Why accuracy fails on imbalanced data?

Because high accuracy may still ignore minority classes.

---

## 2. Precision vs Recall?

* Precision focuses on FP
* Recall focuses on FN

---

## 3. When use F1 Score?

When dataset is imbalanced and both Precision & Recall matter.

---

## 4. ROC-AUC vs Accuracy?

ROC-AUC evaluates classifier quality across thresholds.
Accuracy depends on one threshold.

---

# Python Example

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    confusion_matrix
)

y_true = [1, 0, 1, 1, 0]
y_pred = [1, 0, 0, 1, 1]

print("Accuracy:", accuracy_score(y_true, y_pred))
print("Precision:", precision_score(y_true, y_pred))
print("Recall:", recall_score(y_true, y_pred))
print("F1 Score:", f1_score(y_true, y_pred))
print("Confusion Matrix:")
print(confusion_matrix(y_true, y_pred))
```

````python
import numpy as np
import pandas as pd 
import matplotlib.pyplot as plt 
import sklearn.metrics
````

````python
df = pd.read_csv(r'C:\Users\ipartzix\Downloads\heart.csv')
````

````python
df.head()
````

````python
from sklearn.model_selection import train_test_split
X_train , X_test , y_train , y_test =train_test_split(df.iloc[:,0:-1],df.iloc[:,-1],test_size=0.2,random_state=2)
````

````python
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
````

````python
clf1 = LogisticRegression()
clf2 = DecisionTreeClassifier()
````

````python
clf1.fit(X_train,y_train)
clf2.fit(X_train,y_train)
````

````python
y_pred1 = clf1.predict(X_test)
y_pred2 = clf2.predict(X_test)
````

````python
from sklearn.metrics import accuracy_score,confusion_matrix
print("Accuracy of Logistic Regression",accuracy_score(y_test,y_pred1))
print("Accuracy of Decision Trees",accuracy_score(y_test,y_pred2))
````

````python
confusion_matrix(y_test,y_pred1)
````

````python
print("Logistic Regression Confusion Matrix\n")
pd.DataFrame(confusion_matrix(y_test,y_pred1),columns=list(range(0,2)))
````

````python
print("Decision Tree Confusion Matrix\n")
pd.DataFrame(confusion_matrix(y_test,y_pred2),columns=list(range(0,2)))
````

````python
result = pd.DataFrame()
result['Actual Label'] = y_test
result['Logistic Regression Prediction'] = y_pred1
result['Decision Tree Prediction'] = y_pred2
````
