# Agglomerative Clustering

```python
import numpy as np 
import pandas as pd
import matplotlib.pyplot as plt 
import scipy.cluster.hierarchy as shc

from sklearn.cluster import AgglomerativeClustering 
```

```python
df = pd.read_csv("https://raw.githubusercontent.com/campusx-official/agglomerative-hierarchical-clustering-demo/refs/heads/main/hierarchical-clustering-with-python-and-scikit-learn-shopping-data.csv"
)
print(df.head()) # Genre is not it actually Gender 
```

```python
df.shape
```

```python
df = df.iloc[:, 3:5].values
```

```python
plt.Figure(figsize=(10,7))
plt.title("Customer dendograms")
dend =shc.dendrogram(shc.linkage(df, method="ward"))
```

```python
cluster = AgglomerativeClustering(n_clusters=5, metric="euclidean", linkage="ward")
```

```python
labels_ = cluster.fit_predict(df)
```

```python
labels_
```

```python
plt.figure(figsize=(10, 7))
plt.scatter(df[:, 0], df[:, 1], c=cluster.labels_, cmap="rainbow")
```
