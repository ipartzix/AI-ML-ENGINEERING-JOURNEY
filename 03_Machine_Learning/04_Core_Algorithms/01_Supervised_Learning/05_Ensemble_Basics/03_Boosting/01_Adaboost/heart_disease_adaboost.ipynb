# Heart Disease Dataset

```python
import numpy as np 
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.ensemble import AdaBoostClassifier
from sklearn.model_selection import GridSearchCV
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, classification_report
```

```python
df = pd.read_csv("/home/ipartzix/ML_Dataset/heart.csv")
```

```python
df.sample(10)
```

```python
X = df.drop(columns=["target"])
y = df["target"]
```

```python
# 2. Scale continuous numerical columns
scaler = StandardScaler()
numerical_cols = ["age", "trestbps", "chol", "thalach", "oldpeak"]
X[numerical_cols] = scaler.fit_transform(X[numerical_cols])
```

```python
# 3. One-hot encode integer-based categorical columns
categorical_cols = ["cp", "restecg", "slope", "ca", "thal"]
X_processed = pd.get_dummies(X, columns=categorical_cols, drop_first=True)
```

```python
# 4. Split data into training and test sets
X_train, X_test, y_train, y_test = train_test_split(
    X_processed, y, test_size=0.2, random_state=42, stratify=y
)
```

```python
print(f"Total features after encoding: {X_processed.shape[1]}")
```

```python
X_train
```

```python
adc=AdaBoostClassifier()
adc.fit(X_train,y_train)
```

```python
y_true = np.array(y_test)
y_pred = adc.predict(X_test)
```

```python
accuracy = accuracy_score(y_true, y_pred)
```

```python
print(f"Calculated Test Accuracy: {accuracy:.4f}\n")
print(f"y_true (Actuals):   {y_true[:15]}")
print(f"y_pred (Predicted): {y_pred[:15]}")
```
