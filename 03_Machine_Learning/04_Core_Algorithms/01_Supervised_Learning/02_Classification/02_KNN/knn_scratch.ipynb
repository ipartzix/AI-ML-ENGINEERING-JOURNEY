##  implement knn from scratch 

```python
import numpy as np
from collections import Counter

class knn:
    def __init__(self, k):
        self.n_neighbors = k
        self.x_train = None
        self.y_train = None

    def fit(self, X_train, y_train):
        self.x_train = np.array(X_train)
        self.y_train = np.array(y_train)

    def predict(self, X_test):
        predictions = []
        for i in np.array(X_test):
            distances = []
            for j in self.x_train:
                distances.append(self.calculate_distance(i, j))
            
            # Sort distances and get the indices of the k nearest neighbors
            n_neighbors = sorted(enumerate(distances), key=lambda x: x[1])[:self.n_neighbors]
            pred = self.majority_count(n_neighbors)
            predictions.append(pred)
            
        return np.array(predictions)

    def calculate_distance(self, x1, x2):
        return np.linalg.norm(x1 - x2)

    def majority_count(self, n_neighbors):
        votes = []
        for i in n_neighbors:
            votes.append(self.y_train[i[0]])
            
        # Get the most common class
        return Counter(votes).most_common(1)[0][0]
```
