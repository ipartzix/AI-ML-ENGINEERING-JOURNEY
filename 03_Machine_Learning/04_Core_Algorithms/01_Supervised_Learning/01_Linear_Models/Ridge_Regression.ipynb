# Ridge Regression

Ridge Regression is a regularized version of linear regression that adds a penalty term to reduce model complexity and prevent overfitting.

It adds a penalty term to the ordinary least squares (OLS) cost function.

The Ridge Regression cost function is:


J(θ)=∑i=1n​(yi​−y^​i​)2+λ∑j=1m​θj2​

Where:

- First term → prediction error (RSS)
- Second term → regularization penalty
- λ → regularization parameter
- θj → model weights

## Ridge Regression is Used

Linear Regression can:

- Overfit training data
- Produce unstable coefficients when features are highly correlated

Ridge Regression solves this by:

- Shrinking coefficients
- Reducing model complexity
- Improving generalization on unseen data

## How It Works

Ridge Regression adds an L2 penalty:

λ∑j=1m​θj2​​

Effects:

- Large coefficients become smaller
- Model becomes less sensitive to noise
- Helps prevent overfitting

Unlike Lasso Regression, Ridge Regression:

- Does not make coefficients exactly zero
- Keeps all features in the model

## Key Characteristics
| Feature                   | Ridge Regression         |
| ------------------------- | ------------------------ |
| Regularization Type       | L2                       |
| Reduces Overfitting       | Yes                      |
| Feature Selection         | No                       |
| Handles Multicollinearity | Yes                      |
| Coefficients Become Zero  | No                       |
| Best For                  | Many correlated features |


## Role of Lambda (λ)

The value of λ controls regularization strength.

Small Lambda
- Behaves like Linear Regression
- Less penalty
- Possible overfitting

Large Lambda
- Strong penalty
- Coefficients shrink heavily
- Possible underfitting

## Example

Suppose a model predicts house prices using:

- Area
- Bedrooms
- Floors
- Nearby schools

If these features are highly correlated, Linear Regression may produce unstable weights.

Ridge Regression stabilizes them by shrinking coefficient values.

## Advantages
- Reduces overfitting
- Improves model generalization
- Handles multicollinearity
- Works well with many features
- More stable than OLS

## Disadvantages
- Does not perform feature selection
- Harder to interpret coefficients
- Needs feature scaling before training

## Ridge vs Linear Regression
| Linear Regression              | Ridge Regression          |
| ------------------------------ | ------------------------- |
| No penalty term                | Uses penalty term         |
| Can overfit easily             | Reduces overfitting       |
| Sensitive to multicollinearity | Handles multicollinearity |
| Large coefficients possible    | Coefficients shrink       |


```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
from sklearn.datasets import load_diabetes

data=load_diabetes()
```

```python
print(data.DESCR)
```

```python
X=data.data
y=data.target
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.2,random_state=45)
```

```python
from sklearn.linear_model import LinearRegression
L=LinearRegression()
```

```python
L.fit(X_train,y_train)
```

```python
print(L.coef_)
print(L.intercept_)
```

```python
y_pred=L.predict(X_test)
```

```python
from sklearn.metrics import r2_score,mean_squared_error

print("R2 score",r2_score(y_test,y_pred))
print("RMSE",np.sqrt(mean_squared_error(y_test,y_pred)))
```

```python
from sklearn.linear_model import Ridge
R=Ridge(alpha=100000)
```

```python
R.fit(X_train,y_train)
```

```python
print(R.coef_)
print(R.intercept_)
```

```python
y_pred1=R.predict(X_test)
```

```python
print("R2 score",r2_score(y_test,y_pred1))
print("RMSE",np.sqrt(mean_squared_error(y_test,y_pred1)))
```

```python
m = 100
x1 = 5 * np.random.rand(m, 1) - 2
x2 = 0.7 * x1 ** 2 - 2 * x1 + 3 + np.random.randn(m, 1)

plt.scatter(x1, x2)
plt.show()
```

```python
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import PolynomialFeatures

def get_preds_ridge(x1, x2, alpha):
    model = Pipeline([
        ('poly_feats', PolynomialFeatures(degree=16)),
        ('ridge', Ridge(alpha=alpha))
    ])
    model.fit(x1, x2)
    return model.predict(x1)

alphas = [0, 20, 200]
cs = ['r', 'g', 'b']

plt.figure(figsize=(10, 6))
plt.plot(x1, x2, 'b+', label='Datapoints')

for alpha, c in zip(alphas, cs):
    preds = get_preds_ridge(x1, x2, alpha)
    # Plot
    plt.plot(sorted(x1[:, 0]), preds[np.argsort(x1[:, 0])], c, label='Alpha: {}'.format(alpha))

plt.legend()
plt.show()
```
